```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numeric columns only
numeric_df = seismic_df.select_dtypes(include=[np.number])

# Compute correlations
corr = numeric_df.corr()

# Focus on correlation with your target variables
target_corr = corr[['Predicted Damage Index (01 Scale)', 'Predicted Collapse Probability (%)']].sort_values(
    by='Predicted Damage Index (01 Scale)', ascending=False)

# Display correlation values
print(target_corr.head(26))

# --- Plot the heatmap ---
plt.figure(figsize=(10, 8))
sns.heatmap(target_corr, annot=True, cmap='RdYlBu', vmin=-1, vmax=1)
plt.title('Feature Correlation with Damage Index and Collapse Probability', fontsize=13)
plt.show()
```

Correlation is linear and doesn’t capture interactions therefore the correlation numbers are not the greatest.
To get a more realistic view, a quick tree-based model (like Random Forest) was trained to compute feature importance directly.

```python
from sklearn.ensemble import RandomForestRegressor

# Prepare features (exclude target columns)
X = seismic_df.drop(columns=[
    'Predicted Damage Index (01 Scale)',
    'Predicted Collapse Probability (%)'
])
X = pd.get_dummies(X, drop_first=True)  # Encode categoricals if any

# Target: Damage Index
y = seismic_df['Predicted Damage Index (01 Scale)']
# Fit a quick Random Forest Regressor
rf = RandomForestRegressor(n_estimators=500, random_state=42)
rf.fit(X, y)

# Extract feature importance
importances = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False)

# Plot
plt.figure(figsize=(10, 6))
importances.head(26).plot(kind='barh')
plt.gca().invert_yaxis()
plt.title('Feature Importances for Predicted Damage Index', fontsize=13)
plt.xlabel('Importance Score')
plt.show()
```

```python
y2 = seismic_df['Predicted Collapse Probability (%)']
rf2 = RandomForestRegressor(n_estimators=500, random_state=42)
rf2.fit(X, y2)

importances2 = pd.Series(rf2.feature_importances_, index=X.columns).sort_values(ascending=False)

plt.figure(figsize=(10, 6))
importances2.head(26).plot(kind='barh', color='teal')
plt.gca().invert_yaxis()
plt.title('Feature Importances for Collapse Probability', fontsize=13)
plt.xlabel('Importance Score')
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

# Prepare features (exclude target columns)
X = seismic_df.drop(columns=[
    'Seismic Zone'])
X = pd.get_dummies(X, drop_first=True)  # Encode categoricals if any

# Target: Damage Index
y = seismic_df['Seismic Zone']

# Fit a quick Random Forest Regressor
rf = RandomForestRegressor(n_estimators=500, random_state=42)
rf.fit(X, y)

# Extract feature importance
importances = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False)

# Plot
plt.figure(figsize=(10, 6))
importances.head(27).plot(kind='barh')
plt.gca().invert_yaxis()
plt.title('Feature Importances', fontsize=13)
plt.xlabel('Importance Score')
plt.show()
```

To understand how different structural and seismic features influence building vulnerability, a correlation heatmap was generated, followed by a Random Forest–based feature importance analysis.

The correlation heatmap shows the linear relationships between each numerical feature and the two target variables = Predicted Damage Index (0–1 Scale) and Predicted Collapse Probability (%). Correlation coefficients range from –1 to +1, where positive values indicate that higher feature values are associated with greater damage, while negative values indicate protective effects. In the results, the Predicted Damage Index and Collapse Probability were perfectly correlated (r = 1.0) since the latter was derived from the former. Other features exhibited weak correlations (|r| < 0.1), which is expected because the data were normalized and the targets are nonlinear combinations of multiple factors.

Correlation analysis only captures linear relationships, a Random Forest Regressor was then used to calculate feature importance scores. This model-based method measures how much each feature contributes to reducing prediction error, capturing both linear and nonlinear effects as well as feature interactions.
Together, these analyses provide complementary insights:

The correlation map confirms that the dataset behaves consistently with physical expectations (e.g., stronger ground motion → higher damage).

The feature importance ranking reveals which features have the greatest overall influence in the predictive model, accounting for complex nonlinear interactions that correlation alone cannot detect.

This dual analysis ensures that both statistical and engineering-based dependencies are properly validated before model training, improving interpretability and trust in subsequent regression and classification results.
